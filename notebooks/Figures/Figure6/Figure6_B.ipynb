```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('../../../data/processed/DNA/tmb_per_sample.csv', index_col=0)
```

```python
df1=pd.read_csv('../../../data/processed/DNA/mutation_table.csv', index_col=0)
```

```python
df=df.merge(df1[['General']], left_index=True, right_index=True)
```

```python
sns.violinplot(df, y='total', x='General', hue='General',cut=0)
sns.stripplot(df, y='total', x='General', color='w', linewidth=1)
sns.despine()
#plt.ylabel('TMB - COSMIC cancer driver genes')
plt.ylabel('Tumor mutational burden')
plt.xticks([0,1],['U-DCABP','F-DCABP'])
plt.xlabel('')
plt.tight_layout()
#plt.savefig('TMB_COSMIC_drivers.svg')
```

```python
from scipy.stats import shapiro, ttest_ind, mannwhitneyu

x1 = df[df['General'] == 0]['total'].dropna()
x2 = df[df['General'] == 1]['total'].dropna()
    
stat, pval = mannwhitneyu(x1, x2, alternative="two-sided")
pval
```
